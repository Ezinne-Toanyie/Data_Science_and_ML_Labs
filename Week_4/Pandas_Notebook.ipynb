Pandas library in Python programming enables data manipulation and analysis with structures like Series and DataFrame.

Series is a one-dimensional labelled array capable of holding any datatype


```python
import pandas as pd  #importing the pandas library

# creating a Series from a Python list
# the output datatype is 'object' because the list contains multiple datatype
s = pd.Series([1, 2, 3, 'Numbers', 4, 5])
print(s)
```

DataFrame is a two-dimensional labelled data structure with columns similar to a spreadsheet or SQL table

```python
df = {'name':['Magnus','Rachael','Joanna','King'],
      'gender':['Male','Female','Female','Male'],
      'age':[46,32,29,27],
      'location':['Lagos','Port Harcourt','Ottawa','Boston']}
users = pd.DataFrame(df)
print(users)
```

Data Manipulation

Data in Series and DataFrames can be accessed and manipulated using indexing and slicing

```python
# Index is used to access a single element
print(s[0])
print(users['name'][2])

# Slicing accesses a range of elements
df = pd.DataFrame(users)
df[['name']]
```

Data Combination

This involves joining, merging or concatenating multiple pandas data structure into a single dataset

```python
# Concatenation combines two or more dataframes with similar columns
# either horizontally or vertically
staff_1 = pd.DataFrame({'name':['Peter','Janet','James'],'age':[23,45,32]})
staff_2 = pd.DataFrame({'name':['Lisa','Stacy','Greg'],'age':[42,37,29]})

# Vertical Concatenation(row-wise from top to bottom)
vertical_con = pd.concat([staff_1,staff_2],axis=0,keys=['staff_1','staff_2'])
vertical_con
```

```python
# vertical concatenation by location; first row
vertical_con.loc['staff_1', 0]
```

```python
# vertical concatenation by location; first row,first column
vertical_con.loc['staff_1', 0][0]
```

```python
vertical_con.loc['staff_1', 0]['name':'age']
```

```python
# Vertical Concatenation ignoring the names of the combined dataframes
vertical_con = pd.concat([staff_1,staff_2],axis=0, ignore_index=True)
vertical_con
```

```python
# Horizontal Concatenation(column-wise from left to right)
horizontal_con = pd.concat([staff_1,staff_2],axis=1,keys=['staff_1','staff_2'])
horizontal_con
```

```python
horizontal_con["staff_1"]
```

```python
horizontal_con["staff_2"]
```

```python
horizontal_con["staff_1"][["name"]]
```

MERGE

The merge function is used to combine two or more dataframes based on a common column(s)

```python
df_3 = pd.DataFrame({'Names': ['Ama', 'Barry', 'Celestine', 'Dela'], 'Score1': [1, 2, 3, 4]})
df_4 = pd.DataFrame({'Names': ['Barry', 'Dela', 'Emma', 'Frank'], 'Score2': [5, 6, 7, 8]})
```

```python
# keeps records of 'Names' common to both dataframes
pd.merge(df_3, df_4, on='Names', how='inner')
```

```python
# keeps all records from both dataframes
pd.merge(df_3, df_4, on='Names', how='outer')
```

```python
# keeps all records on the left dataframe and common records from the right df
pd.merge(df_3, df_4, on='Names', how='left')
```

```python
# keeps all records on the right dataframe and common records from the left df
pd.merge(df_3, df_4, on='Names', how='right')
```

```python
# Merge instance on two dataframes having different column names
df_5 = pd.DataFrame({'Name1': ['Ama', 'Barry', 'Celestine', 'Dela'], 'Score': [1, 2, 3, 4]})
df_6 = pd.DataFrame({'Name2': ['Barry', 'Dela', 'Emma', 'Frank'], 'Score': [5, 6, 7, 8]})
merged_df = pd.merge(df_5, df_6, left_on='Name1', right_on='Name2', how='inner', suffixes=('_df5', '_df6'))
merged_df
```

```python
# Merge instance on two dataframes having different column names
df_5.merge(df_6,left_on='Name1', right_on='Name2', how='inner', suffixes=('_df5', '_df6'))
```

JOIN

The join() function is used to combine two dataframes on their indexes

```python
df_7 = pd.DataFrame({'value1': [1, 2, 3, 4]}, index=['A', 'B', 'C', 'D'])
df_8 = pd.DataFrame({'value2': [5, 6, 7, 8]}, index=['B', 'D', 'E', 'F'])
df_7.join(df_8, how = "outer")
```

Filtering: Data can be filtered based on conditions.

```python
df = {'name':['Magnus','Rachael','Joanna','King'],
      'gender':['Male','Female','Female','Male'],
      'age':[46,32,29,27],
      'location':['Lagos','Port Harcourt','Ottawa','Boston']}
users = pd.DataFrame(df)
filtered_users = users[users['age'] < 40]
print(filtered_users)
```

Adding and Removing Columns

```python
data = {'Name': ['John', 'Alice', 'Bob'],
        'Age': [30, 25, 35],
        'City': ['New York', 'Paris', 'London']}
df = pd.DataFrame(data)
print(df)

# Adding a new column
df['Gender'] = ['Male', 'Female', 'Male']
print(" ")    #spacing
print(df)

# Removing a column
df.drop('City', axis=1, inplace=True)
print(" ")    #spacing
df

```

Grouping and Aggregation

Data can be grouped based on one or more columns and have aggregation functions perfomed on them.

```python
# Creating a DataFrame from a dictionary
data = {'Num': [10, 20, 10, 20, 20, 30, 30],
        'Age': [30, 30, 25, 25, 25, 25, 35],
        'val': [50, 52, 64, 20, 50, 60, 70]}
df = pd.DataFrame(data)
print(df)
```

```python
# Grouping and aggregating
grouped_mean = df.groupby('Age').mean()
grouped_mean
```

```python
max_age = df.groupby('Age').max()
max_age
```

dropna()

Reading CSV Files:

*   Use the read_csv() function to read data from a CSV (Comma-Separated Values) file.


**Syntax**

Options:

sep: Specifies the delimiter (default is ,).

header: Specifies the row number to use as column names.

index_col: Specifies the column to use as the row index.

usecols: Specifies a subset of columns to read.
