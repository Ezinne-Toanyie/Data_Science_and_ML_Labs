## Numpy is a general purpose array-processing package in python.

It mostly deals with numbers arranged in arrays

```python
import numpy as np
```

```python
# 1-dimensional array from a python list
arr1d = np.array([1,2,3,4,5])
arr1d
```

```python
# 2-dimensional array from a nested python list; presents in rows and columns
arr2d = np.array([[1,2,3],[4,5,6]])
arr2d
```

```python
# an array of 0s; 3 rows and 4 columns
zeros_arr = np.zeros((3,4))
print(zeros_arr)
```

## Array attributes


```python
# Shape
shape = arr2d.shape  # returns the number of rows and columns in the array
print(shape)

zeros_arr.shape
```

```python
# Size
size = arr2d.size   # returns the number of elements in the array
print(size)

zeros_arr.size
```

```python
# Data type
data_type = arr2d.dtype
print(data_type)

zeros_arr.dtype
```

## Array Operations

```python
arr2d
```

```python
# Addition
addition = 2 + arr2d    # adding 2 to all the elements of the array
print(addition)
```

```python
# Subtraction
subtraction = arr2d - 3 # subtract 3 from all the elements in the array
print(subtraction)
```

```python
# Multiplication: multiplies through all the elements in the array
multiplication = 4 * arr2d
print(multiplication)
```

## Aggregation Functions

```python
# Sum: adds up all the elements in the array
total_sum = arr2d.sum()
print(total_sum)
```

```python
# Max: maximum value of all the elements in the array
max_val = arr2d.max()
print(max_val)
```

```python
# Min: minimum value of all the elements in the array
min_val = arr2d.min()
print(min_val)
```

## Indexing

```python
# return rows with index 0,stop before index 1, and all columns
first_row = arr2d[0:1,:]
print(first_row)
```

```python
# return row from index 1 to the last and all columns
second_row = arr2d[1:,:]
print(second_row)
```

```python
# return rows with index 0,stop before index 1, and columns from index 0, stop
# before index 2
first_row_element = arr2d[0:1,0:2]
print(first_row_element)
```

```python
# return rows from index 1 and columns from index 1
second_row_element = arr2d[1:,1:]
print(second_row_element)
```

## Broadcasting

Broadcasting in NumPy allows operations between arrays of different shapes by automatically adjusting their dimensions to be compatible.
Broadcasting enables operations between arrays of different shapes without requiring explicit reshaping.

```python
# automatically stretches the single number so it can be added to each item in
# the list
a = np.array([1,2,3])
b = 4
print(a+b)
```

```python
a = np.array([1,2,3])
b = 6
print(a*b)
```

## Random Number Generation

```python
# random.rand() function generates random numbers between 0 and 1
random_numbers = np.random.rand(3,3)
print(random_numbers)
```

```python
# random.randint() function generates random integers within specified range
# including the lower boundary but excluding the upper boundary
random_numbers_ = np.random.randint(0,100,6)
print(random_numbers_)
```

```python
# reshaping generated random numbers
random_numbers_ = np.random.randint(0,100,6).reshape(3,2)
print(random_numbers_)
```

```python
# random.randn() returns random numbers that fall within a normal distribution
random_numbers1 = np.random.randn(4, 6)
print(random_numbers1)
```

```python
items = np.array (["A", "B", "C", "D"])
# random.choice() randomly picks elements from the array
random_numbers2 = np.random.choice(items, 3)
print(random_numbers2)
```

## Linear Algebra

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
```

```python
# matrix addition
print(A + B)
```

```python
# matrix subtraction
print(A - B)
```

```python
# matrix multiplication (Element-wise multiplication)
print(A * B)
```

```python
# matrix multiplication
print(A @ B)
```

```python
result = np.matmul(A,B)
print(result)
```

```python
# Standard Matrix multiplication
F = np.dot(A,B)
print(F)
```

```python
A = np.array([[5,6,7],[8,9,0]])
A
# Transpose of a matrix
transpose_A = np.transpose(A)
print(transpose_A)
```

```python
B = np.array([[1,2],[3,4]])
# Inverse of a matrix
B_inv = np.linalg.inv(B)
print(B_inv)
```

```python
# identity matrix
identity_matrix = np.eye(6)
print(identity_matrix)
```

```python
# determinant of a matrix: a single numerical value computed by a square matrix
determinant = np.linalg.det(B)
print(int(determinant))
```

## Deleting elements in an array

```python
arr = np.array([1,2,3,4,5])
arr_deleted = np.delete(arr,2)
print(arr_deleted)
```

```python
arr = np.array([1,2,3,4,5])
m = arr != 5
new_arr = arr[m]
print(new_arr)
```

## Statistical Functions

```python
arr = np.array([11,11,12,32,41,50,7,])
```

```python
# mean
arr_mean = np.mean(arr)
print(arr_mean)
```

```python
# median
arr_median = np.median(arr)
print(arr_median)
```

```python
# mode : numpy does not have the mode method, so scipy can be used
arr = np.array([11,11,12,32,11,41,50,7,])
from scipy import stats
arr_mode = stats.mode(arr,keepdims = False)
print(arr_mode)
```

```python
# standard deviation
arr_std = np.std(arr)
print(arr_std)
```
