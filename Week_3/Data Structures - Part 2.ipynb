ARITHMETIC OPERATORS

```python
# ADDITION:returns the sum of values
X = 10
Y = 15
print (X + Y)
```

```python
# SUBTRACTION: returns the difference between values
Y-X
```

```python
# MULTIPLICATION: returns the product of values
X * Y
```

```python
# DIVISION(/): performs standard division
# always returns a float even if the result is a whole number with no remainders
x = 300
y = 6
x/y
```

```python
# floor division(//) : performs division but ignores remainders
# returns an integer

x=300
y=7
x//y
```

```python
# modulo(%): returns remainder after division
x%y
```

```python
# exponential: returns the result of x raised to the power of y
x = 5
y = 3
x**y
```

BITWISE OPERATORS

In Python, bitwise operations are carried out on the binary equivalent of numeric data


```python
# where the  prefix 0b means binary
print (bin(10))
print (bin(3))
print (bin(2))
```

```python
# & is the symbol for AND
x = 10
y = 3
x & y
```

```python
# | is the symbol for OR
x | y
```

```python
# ^ is the symbol for XOR; exclusive OR
x ^ y
```

```python
~x # is the symbol for bitwise negation
```

```python
# shifts the bits of a number to the right by a specified number of positions.
x >> 2
```

```python
# shifts the bits of a number to the left by a specified number of positions.
x << 2
```

MEMBERSHIP OPERATORS

```python
# Python has two membership operators; 'in' and 'not in'
# They check if a value is found in a sequence
numbers = [1,2,3,4,5]
grade = "Excellent"
print(6 in numbers)
print('w' not in grade)
```

IDENTITY OPERATORS

```python
# Python has two identity operators; 'is' and 'is not'
# They check if two objects are the same in memory
# not just equal in value
a = [1,2,3,4,5]
b = [1,2,3,4,5]
print(a is b)
print(a is not b)
```
