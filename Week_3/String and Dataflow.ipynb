```python
# STRING: a sequence of characters used to represent text
Name_ = "ParoCyber LLC"
Name_
```

```python
# upper() converts to uppercase
Name_ = Name_.upper()
Name_
```

```python
# lower() converts to lowercase
Name_ = Name_.lower()
Name_
```

```python
# isalpha() checks if all the characters are alphabets
Name_ = Name_.isalpha()
Name_
```

```python
# isdigit() checks if all the characters are numbers
Name_ = Name_.isdigit()
Name_
```

```python
# islower() checks if all the characters are in lowercase
Name_ = Name_.islower()
Name_
```

```python
# isupper() checks if all the characters are in uppercase
Name_ = Name_.isupper()
Name_
```

```python
# split() separates texts by delimiters
Flowers = "Flowers are beautiful"
word = Flowers.split(" ")
word
```

```python
# join()
Flowers = "Flowers are beautiful"
"_".join(word)
```

```python
# replace
Flowers = "Flowers are beautiful"
Flowers.replace("beautiful", "lovely")
```

```python
# startswith() shows what an element starts with
Flower = "Hibiscus"
print (Flower.startswith("b"))
print (Flower.startswith("H"))
```

```python
# endswith() shows what an element ends with
Flower = "Hibiscus"
print (Flower.endswith("S"))
print (Flower.endswith("s"))
```
